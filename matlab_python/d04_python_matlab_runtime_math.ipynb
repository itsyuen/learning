#### Exercise 5 Loop control
- if and elif
- while
- conditions

compared with matlab, the if/else-if/while are similar with a little difference

```python
# check numbers list

numbers = [1, 2, 3, 4]

print("Current numbers list are:")
for i, number in enumerate(numbers):
    print(f"Index: {i} has value: {numbers[i]}")    # can also be {number} as it is called
    print(f"And the type of object is: {type(number)}" )

numbers = [float(value) for value in numbers]

print(f"now we made the {numbers} list into a {type(numbers)}")  # type casted

numbers[2] = -3.0

print(f" we change one of the values directly")
numbers

for number in numbers:      # using if/else if inside for loop to check on each elemental object
    if number > 0.0:
        print("its a positive number!")
    elif number < 0.0:
        print("its a negative number!")
    else:
        print("it should be zero or something else?")

print(numbers)

for number in numbers:
    if number < 0:
        number = -1*number      # reassings 'number'; does not change numbers[i], though float is immutable 

print(numbers)      # has this changed the list object?

for i in range(len(numbers)):   # one way of using index to change object, calling index
    if numbers[i] < 0:
        numbers[i] = -1*numbers[i]

print(numbers)

for i, number in enumerate(numbers):    # another way to change the object, having both index and value object
    if number > 0:
        numbers[i] = -number    # notice using current object to mutate the list object

print(numbers)
```
