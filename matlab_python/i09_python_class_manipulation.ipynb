# Classes, Object State, and Callback-Driven Programming

## Objective

This lecture exercise prepares students for `rclpy` and ROS 2 node
construction by developing the programming-language literacy needed to
understand:

- classes as containers for persistent data/state and behavior
- objects as instances of classes
- functions and class methods
- callback-driven execution
- strongly typed data and ROS 2 messages
- inheritance and ROS 2 `Node` classes
- program entry and node execution

The objective is not to provide a complete object-oriented programming
course. Instead, C++, Python, and MATLAB examples are used to understand
the software structures that later appear naturally in ROS 2.

The central engineering question is:

> How do we organize computation and state inside a component, and how
> does typed data cross the boundary between components?
