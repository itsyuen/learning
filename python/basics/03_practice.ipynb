# Python Lists vs NumPy Arrays in ROS 2 Development

One of the biggest differences between MATLAB and Python is that Python separates **general programming objects** from **numerical computation**.

## Rule of Thumb

> **Python Lists are for "things".**
>
> **NumPy Arrays are for "numbers".**

| Task | Python List | NumPy Array | Reason |
|:-----|:-----------:|:-----------:|:------|
| Store filenames | ✅ | ❌ | Filenames are strings (objects) |
| Store folder contents (`os.listdir`) | ✅ | ❌ | Returns a list of filenames |
| Store ROS node names | ✅ | ❌ | Node names are strings |
| Store topic names | ✅ | ❌ | Topics are strings |
| Store robot links / objects | ✅ | ❌ | Robot components are Python objects |
| Store ROS publishers/subscribers | ✅ | ❌ | Publishers are objects |
| Store JointState names | ✅ | ❌ | `msg.name` is a list of strings |
| Store robot state (q, q_dot, q_ddot) | ❌ | ✅ | Numerical vectors |
| Joint velocity vector | ❌ | ✅ | Vector computation |
| Joint acceleration vector | ❌ | ✅ | Vector computation |
| Jacobian matrix | ❌ | ✅ | Matrix algebra |
| Rotation matrix | ❌ | ✅ | Linear algebra |
| Homogeneous transform | ❌ | ✅ | Matrix multiplication |
| Inertia matrix M(q) | ❌ | ✅ | Matrix operations |
| Coriolis matrix C(q,q̇) | ❌ | ✅ | Matrix operations |
| Gravity vector G(q) | ❌ | ✅ | Numerical computation |
| EKF covariance matrix | ❌ | ✅ | Matrix algebra |
| Robot trajectory | Usually converted to NumPy | ✅ | Efficient numerical computation |
| Camera image | ❌ | ✅ | Pixel values are numbers |
| Point cloud | ❌ | ✅ | Large numerical dataset |
| Machine learning tensors | ❌ | Usually NumPy (or PyTorch) | Numerical computation |

# Typical ROS 2 Workflow

```text
Operating System
        │
        ▼
Python List
(os.listdir(), node names, topic names)
        │
        ▼
ROS Messages
(JointState, Pose, Twist)
        │
        ▼
Convert numerical data
        │
        ▼
NumPy Array
(q, q_dot, J, M(q), ...)
        │
        ▼
Robot Mathematics
        │
        ▼
Controller Output
        │
        ▼
Publish ROS Message
```

# Typical ROS 2 Example

```python
import os
import numpy as np

# Programming objects
files = os.listdir("images")

# ROS message data
joint_names = ["joint1", "joint2", "joint3"]

# Numerical computation
q = np.array([0.2, 0.4, -0.1])
qd = np.array([0.1, 0.0, 0.3])

J = np.eye(3)

xdot = J @ qd
```

Notice that the same program naturally contains both **lists** and **NumPy arrays**, each serving a different purpose.

# Summary

| Think of... | Use... |
|-------------|---------|
| Files | Python List |
| Strings | Python List |
| ROS Objects | Python List |
| Messages | Python List (contains objects) |
| Vectors | NumPy Array |
| Matrices | NumPy Array |
| Dynamics | NumPy Array |
| Linear Algebra | NumPy Array |

## Golden Rule

> **Python Lists organize your robotics software.**
>
> **NumPy Arrays perform your robotics mathematics.**
